## Задание 1. Арифметические операции, ввод/вывод, форматируемые строки (0,5 балла)

**Задача:**

Напишите программу, которая:
1. Запрашивает у пользователя два целых числа.
2. Выполняет над этими числами 6 арифметических операций (+, -, *, /, %, //).
3. Выводит с помощью форматируемых строк результаты выполнения операций в следующем виде:<br>
`6 + 13 = 19`<br>
`6 - 13 = -7`<br>
`6 * 13 = 78` <br>
и т.д.<br>
Вещественные значения должны выводиться с точностью 3 знака после десятичной точки.

```python
# Место для решения задания 1
a = int(input("Введите число: "))
b = int(input("Введите число: "))
print(f"{a} + {b} = {a+b}")
print(f"{a} - {b} = {a-b}")
print(f"{a} * {b} = {a*b}")
print(f"{a} / {b} = {a/b:.3f}")
print(f"{a} % {b} = {a%b}")
print(f"{a} // {b} = {a//b}")
```

## Задание 3. Цикл с предусловием (1,5 балла)

**Задача:**

Управляемый РОБОТ тестируется в прямоугольном помещении размером N х M метров (N, M >= 10).  Начальное положение РОБОТА – точка с координатами (x,y). Оператор управляет РОБОТом командами:
- север - N,  
- юг - S,  
- запад - W,
- восток - E,
- закончить -  X.  
После каждой команды РОБОТ перемещается на 1 метр в нужную сторону, и программа сообщает его новую позицию. Если РОБОТ упёрся в стену, то он не должен пытаться перемещаться в сторону стены, в этом случае его позиция не меняется. Нумерация идет снизу вверх и слева направо.

Пример (размер комнаты 10х10):  
`[Программа]: Введите исходные координаты Робота (x,y):`  
`[Оператор]: 5`  
`[Оператор]: 9 `  
`[Программа]: Марсоход находится на позиции 5, 9, введите команду:`  
`[Оператор]: N`  
`[Программа]: Марсоход находится на позиции 5, 10, введите команду:`  
`[Оператор]: N`  
`[Программа]: Марсоход находится на позиции 5, 10, введите команду:`  
`[Оператор]: W`  
`[Программа]: Марсоход находится на позиции 4, 10, введите команду:`  
`[Оператор]: X`  


```python
# Место для решения задания 3
max_x = int(input("Максимальная координата х: "))
max_y = int(input("Максимальная координата y: "))

print("[Программа]: Введите исходные координаты Робота (x,y)")
x = int(input("Координата х"))
y = int(input("Координата y"))
print(f"[Оператор]: {x}")
print(f"[Оператор]: {y}")

while True:
    print(f"[Программа]: Маскоход находится на позиции {x}, {y}, введите команду")
    command = input("Направление: ")
    if command == "N":
        print("[Оператор]: N")
        if y < max_y:
            y += 1
    elif command == "S":
        print("[Оператор]: S")
        if y > 0:
            y -= 1
    elif command == "W":
        print("[Оператор]: W")
        if x > 0:
            x -= 1
    elif command == "E":
        print("[Оператор]: E")
        if x < max_x:
            x += 1
    elif command == "X":
        print("[Оператор]: X")
        break
```

## Задание 4. Цикл `for` и `while`(1 балл)

**Задача:**

С клавиатуры вводится N целых чисел. Если вводимое число имеет сумму цифр больше 10, то вывести это число и его сумму цифр. Посчитать, сколько таких чисел получилось.
**Не использовать функции и списки!**




```python
# Место для решения задания 4
n = int(input("Количество чисел: "))
count = 0

for i in range(n):
    sm = int(input(f"Введите число: "))
    t = sm
    s = 0

    while t > 0:
        s += t % 10
        t = t // 10
    if s > 10:
        print(f"Число: {sm} Сумма цифр: {s}")
        count += 1

print(f"Количество чисел с суммой больше 10: {count}")
```

## Задание 5. Комбинированное: угадай число (1 балл)

**Задача:**

Пользователь загадывает случайное число от 1 до 100.
Компьютер пытается угадать. Пользователь:
- Даёт подсказки: "Больше", "Меньше".
- Если компьютеру угадал число, то пишет "Угадано" и цикл должен завершаться (`break`).

Компьютер должен:
- Считать количество попыток.
- После угадывания выводит число попыток.
Количество попыток компьютера не длжно превышать 7.

```python
l = 1
r = 100
count = 0
print("Загадайте число от 1 до 100.")

while True:
    k = (l + r) // 2
    count += 1

    print(f"Попытка {count}. Ваше число: {k}?")
    answer = input("Больше/Меньше/Угадано: ")

    if answer == "Угадано":
        print(f"Угадано. Количество попыток: {count}")
        break
    elif answer == "Больше":
        l = k + 1
    elif answer == "Меньше":
        r = k - 1
```
