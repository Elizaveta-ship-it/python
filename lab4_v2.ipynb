## Задание 1. Генераторы и итераторы (1 балл)

**Темы:** итераторы и генераторы .


**Что нужно сделать:**
Напишите функцию-генератор fibonacci(n), которая генерирует первые n чисел Фибоначчи.
Затем создайте итератор evenIndexIterator, который принимает любую последовательность и возвращает только элементы, стоящие на чётных индексах.  
Проверьте работу программы на следующих примерах:   
- для Фибоначчи на последовательности `[0,1,2,5,8,15]`  
- для итератора на последовательностях `[10, 20, 30, 40, 50], ['a', 'b', 'c', 'd', 'e', 'f', 'g', 'h', 'i'], [1, "two", 3.0, [4, 5], (6, 7), {8: 9}, True, None]`      


```python
import types

def fibonacci(n):
    a, b = 0, 1
    count = 0
    while count < n:
        yield a
        a, b = b, a + b
        count += 1

class e:
    def __init__(self, sequence):
        self.sequence = sequence
        self.index = 0
    def __iter__(self):
        return self
    def __next__(self):
        if self.index >= len(self.sequence):
            raise StopIteration
        value = self.sequence[self.index]
        self.index += 2
        return value

print(list(fibonacci(6)))

seq1 = [10, 20, 30, 40, 50]
print(list(e(seq1)))

seq2 = ['a', 'b', 'c', 'd', 'e', 'f', 'g', 'h', 'i']
print(list(e(seq2)))

seq3 = [1, "two", 3.0, [4, 5], (6, 7), {8: 9}, True, None]
print(list(e(seq3)))
```

## Задание 2. Модуль NumPy (1 балл)

**Темы:** работа с одномерным массивом

**Что нужно сделать:**

Создайте массив из 20 случайных целых чисел от -100 до 100. Выполните следующие действия:
- Найдите минимальное, максимальное и среднее значение.  
- Получите массив из отрицательных элементов.
- Замените все значения, больше 10, на 0.
- Вычислите среднее аримфметическое элементов, находящихся на нечётных позициях.

```python
import numpy as np

arr = np.random.randint(-100, 101, size=20)
print(f"Исходный массив: {arr}")

min_val = np.min(arr)
max_val = np.max(arr)
mean_val = np.mean(arr)
print(f"Минимальное значение: {min_val}")
print(f"Максимальное значение: {max_val}")
print(f"Среднее значение: {mean_val:.2f}")

n = arr[arr < 0]
print(f"Массив отрицательных элементов: {n}")

r = arr.copy()
r[r > 10] = 0
print(f"Массив после замены значений > 10 на 0: {r}")

o = arr[1::2]
m = np.mean(o)
print(f"Среднее арифметическое на нечетных позициях: {m:.2f}")
```

## Задание 4. Работа с датой и временем (2 балла)

**Темы:** Модуль datetime

**Что нужно сделать:**
Программа просит пользователя ввести дату своего рождения в формате ДД.ММ.ГГГГ (например, 15.08.2003). Нужно определить:  
- Какой это был день недели (название на русском или английском).  
- Был ли этот год високосным.  
- Сколько пользователю полных лет на текущий момент (с учетом прошедших дней рождения).  

Пример ожидаемого вывода:

`Введите дату рождения (дд.мм.гггг): 15.08.2003`  
`Вы родились в пятницу.`  
`2003 год не был високосным.`  
`Сейчас вам 20 полных лет.`  

```python
from datetime import datetime as dt
import calendar as cal

s = input("Введите дату рождения (дд.мм.гггг): ")
d = dt.strptime(s, "%d.%m.%Y")
n = dt.now()

days = ["понедельник", "вторник", "среду", "четверг", "пятницу", "субботу", "воскресенье"]
print(f"Вы родились в {days[d.weekday()]}.")

if cal.isleap(d.year):
    print(f"{d.year} год был високосным.")
else:
    print(f"{d.year} год не был високосным.")

age = n.year - d.year
if (n.month < d.month) or (n.month == d.month and n.day < d.day):
    age -= 1

print(f"Сейчас вам {age} полных лет.")
```
