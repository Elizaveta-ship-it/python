## Задание 1. Списки и list comprehension (1 балл)

**Темы:** списки, методы списков, list comprehension.


**Что нужно сделать:**
Напишите программу, которая выполняет следующие действия:
1. Генерирует случайным образом список L1 из 20 элементов в диапазоне от -100 до 100 и выводит его на экран,
2. Формирует на основе этого списка новый список L2, в котором все четные элементы будут уменьшены в 2 раза, а нечетные элементы увеличены в 2 раза, и выводит его на экран (обязательно использовать List comprehension),
3. Вставляет после каждого отрицательного элемента списка L1 число 0 и выводит список на экран (нельзя использовать дополнительный список!),
4. Удаляет из списка L1 все элементы, которые лежат между значениями K1 и K2, и выводит его на экран (нельзя использовать дополнительный список!),
5. Переносит первые три элемента списка L1 в конец списка и выводит список на экран (нельзя использовать дополнительный список!),
6. Сортирует список L1 по убыванию

```python
# Ваш код здесь
import random

print("Список из рандомных значений")
L1 = [random.randint(-100,100) for i in range(20)]
print(L1)

print("Умножили/поделили")
L2 = [x//2 if x%2 == 0 else x*2 for x in L1]
print(L2)

print("Список с нулями")
for i in range(len(L1)):
    if L1[i] < 0:
        L1.insert(i+1, 0)
print(L1)

K1 = int(input("Введите значение K1: "))
K2 = int(input("Введите значение K2: "))
if K1 > K2:
    K1, K2 = K2, K1
print(f"Элементы от {K1} до {K2} будут удалены")

for i in range(len(L1) -1, -1, -1):
    if K1 < L1[i] < K2:
        del L1[i]
print(L1)

print("Перенесли три значения")
if len(L1) >= 3:
    L1 = L1[3:] + L1[:3]
print(L1)

print("Список по убыванию")
L1.sort(reverse=True)
print(L1)
```

## Задание 2. Кортежи. Строки (1 балл)

**Темы:** кортежи, строки, срезы строк.

**Что нужно сделать:**
1. Создайте кортеж из 10 произвольных слов.
2. Напишите программу для игры "Анаграммы", которая будет случайным образом извлекать из кортежа слово, перемешивать буквы в этом слове и предлагать пользователю угадать заданное слово. На угадывание слова давать 3 попытки.


```python
# Ваш код здесь
import random

w = ("испанцы", "кирпич", "бестолочь", "телевизор", "квартира", "интернет", "стена", "система", "часовня", "трактор")

print("Игра 'Анаграммы'")

secret = random.choice(w)

letters = list(secret)
random.shuffle(letters)
shuffled = "".join(letters)

print(f"Перемешанное слово: {shuffled}")

a1 = 3
n = False

for a2 in range(1, a1 + 1):
    answer = input(f"Попытка {a2} из {a1}. Ваш ответ: ")


    if answer == secret:
        print("Угадано")
        n = True
        break
    else:
        if a2 < a1:
            print("Неверно. Попробуйте еще раз.")
        else:
            print("Неверно.")

if not n:
    print(f"Попытки закончились. Загаданное слово было: {secret}")
```

## Задание 3. Функции  (3 балла)

**Темы:** Функции: создание, параметры возврат значений, docstring

Вам необходимо написать несколько функций для обработки среднемесячных температур

```python
temp = [-14, -16, -4, 1, 10, 16, 19, 18, 15, 2, -2, -13]
```

**Задачи:**
1. Функция temp_statistics(temp)
Принимает список среднемесячных температур. Возвращает кортеж из 4 значений:
- средняя температура за год (округлённая до 1 знака)
- максимальная температура
- минимальная температура
- количество значений в списке

Добавьте docstring с описанием параметра и возвращаемого значения.

2. Функция filter_temp(temp, threshold, above=True)
Принимает список температур, пороговое значение threshold и флаг above.
Если above = True — возвращает список температур выше или равных threshold.
Если above = False — возвращает список температур ниже threshold.
Значение по умолчанию для above = True.  

Добавьте docstring.

3. Функция farengeit(temp)
Принимает список температур.
Возвращает новый список, где к каждой температуры из градусов Цельсия переведены в Фаренгейты.
Формула: F = C * 9/5 + 32
Используйте list comprehension внутри функции.

Добавьте docstring.

Напишите код, который:
1. Вызывает temp_statistics и выводит статистику.
2. Вызывает filter_temp c параметрами threshold=2, above=False, выводит список температур ниже 2
3. Вызывает farengeit и выводит новый список температур.


```python
# Ваш код здесь
def temp_statistics(temperatures):
    """
    Принимает список среднемесячных температур
    Возвращает кортеж из 4 значений:
    a: средняя температура за год
    mx: максимальная температура
    mn: минимальная температура
    count: количество значений в списке
    """
    a = round(sum(temperatures) / len(temperatures), 1)
    mx = max(temperatures)
    mn = min(temperatures)
    count = len(temperatures)
    return a, mx, mn, count

def filter_temp(temperatures, threshold, above=True):
    """
    Фильтрует список температур на основе порогового значения
    Параметры:
    temperatures: Список температур
    threshold: Пороговое значение
    above: если True — возвращает температуры выше или равные threshold
           если False — возвращает температуры ниже threshold
    Возвращает отфильтрованный список температур
    """
    if above:
        return [t for t in temperatures if t >= threshold]
    else:
        return [t for t in temperatures if t < threshold]

def farengeit(temperatures):
    """
    Принимает список температур в градусах Цельсия
    Возвращает новый список температур в градусах Фаренгейта
    Формула: F = C * 9/5 + 32
    """
    return [(c * 9 / 5) + 32 for c in temperatures]

temp = [-14, -16, -4, 1, 10, 16, 19, 18, 15, 2, -2, -13]

print("Статистика температур:")

a2, mx2, mn2, count = temp_statistics(temp)
print(f"1. Средняя температура за год: {a2}")
print(f"2. Максимальная температура: {mx2}")
print(f"3. Минимальная температура: {mn2}")
print(f"4. Количество значений в списке: {count}")

print("\nТемпературы ниже 2:")
cold = filter_temp(temp, threshold=2, above=False)
print(f"{cold}")

print("\nТемпературы в Фаренгейтах:")
fahrenheit_temps = farengeit(temp)
print(f"{fahrenheit_temps}")
```
